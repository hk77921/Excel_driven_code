```python
# Initialize broker
load_dotenv()
api_key = os.getenv("KITE_API_KEY")
access_token = os.getenv("KITE_ACCESS_TOKEN")
print(f"Using API Key: {api_key}")
print(f"Using Access Token: {access_token}")    

if not api_key or not access_token:
    raise ValueError("Missing credentials in .env!")
kite = KiteConnect(api_key=api_key)
kite.set_access_token(access_token)
```

```python
historical_data = kite.historical_data(instrument_token=256265,from_date=datetime(2025, 12, 18),to_date=datetime.today(), interval="5minute")
```

```python
historical_data = pd.DataFrame(historical_data)
historical_data
```

Get the Hightest High and lowest Low

```python
historical_data.describe()
```

```python
#get the Hightest High and lowest Low for the dataframe
highest_high = historical_data['high'].max()
lowest_low = historical_data['low'].min()
print(f"Highest High: {highest_high}")
print(f"Lowest Low: {lowest_low}")
```

```python
SortedData=historical_data.sort_values(by='close')

SortedData.head(3)
```

```python
# Calulate SMA and EMA
historical_data['SMA_20'] = ta.sma(historical_data['close'], length=20)
historical_data['EMA_20'] = ta.ema(historical_data['close'], length=20)
historical_data.tail(3)
```

```python
#calculate RSI
historical_data['RSI_14'] = ta.rsi(historical_data['close'], length=14)
historical_data.tail(3)
```

```python
#super trend
supertrend = ta.supertrend(historical_data['high'], historical_data['low'], historical_data['close'], length=7, multiplier=3.0)

#add supertrend to historical_data
historical_data = pd.concat([historical_data, supertrend], axis=1)
historical_data
#full code above
```

```python
#candelstick patterns
historical_data['CDL_DOJI'] = ta.cdl_doji(historical_data['open'], historical_data['high'], historical_data['low'], historical_data['close'])
historical_data[['open', 'high', 'low', 'close', 'CDL_DOJI']].tail(3)
```

```python
#candelstick  patterns - Engulfing pattern


historical_data['CDL_ENGULFING'] = ta.cdl_inside(historical_data['open'], historical_data['high'], historical_data['low'], historical_data['close'])

historical_data.tail(3)
```
